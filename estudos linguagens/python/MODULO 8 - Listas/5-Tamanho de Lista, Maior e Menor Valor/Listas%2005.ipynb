# Algumas Funções Básicas de Lista

## Tamanho da Lista

tamanho = len(lista)

```python
produtos = ['apple tv', 'mac', 'iphone x', 'IPad', 'apple watch', 'mac book', 'airpods']
```

- Quantos produtos temos a venda?

```python
tamanho = len(produtos)
print(f'Temos {tamanho} produtos')
```

## Maior e Menor Valor

maior = max(lista)

menor = min(lista)

```python
vendas = [1000, 1500, 15000, 270, 900, 100, 1200]
```

- Qual o item mais vendido?
- Qual o item menos vendido?

```python
mais_vendido = max(vendas)
menos_vendido = min(vendas)
print(f'O produto mais vendido teve {mais_vendido} vendas e o menos vendido teve {menos_vendido} vendas')
```

```python
i = vendas.index(mais_vendido) #ele ta me dando a posição do mais vendido
produto_mais_vendido = produtos[i] #ele relaciona a posição de duas listas
print(f' o produto mais vendido foi {produto_mais_vendido} com {mais_vendido} vendas')

i = vendas.index(menos_vendido) #ele ta me dando a posição do mais vendido
produto_menos_vendido = produtos[i] #ele relaciona a posição de duas listas
print(f' o produto menos vendido foi {produto_menos_vendido} com {menos_vendido} vendas')
```
