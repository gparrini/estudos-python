```python
#correção com gemini

def analise_metas(meta, dic):
    # 1. Preparamos a "cesta" vazia ANTES de começar a checar os vendedores
    vendedores_meta = []
    
    # 2. Usamos .items() para pegar o nome (vendedor) e o número (vendas) juntos
    for vendedor, vendas in dic.items():
        if vendas >= meta:
            # Se bateu a meta, joga o nome na cesta
            vendedores_meta.append(vendedor)
            
    # 3. O loop acabou! Agora calculamos a porcentagem do lado de fora.
    # len(vendedores_meta) = quantos bateram a meta
    # len(dic) = total de vendedores participando
    porcent_meta = len(vendedores_meta) / len(dic)
    
    # Entregamos as duas respostas
    return vendedores_meta, porcent_meta

# Testando (não precisa mudar essa parte)
meta = 10000
vendas = {
    'João': 15000,
    'Julia': 27000,
    'Marcus': 9900,
    'Maria': 3750,
    'Ana': 10300,
    'Alon': 7870,
}

print(analise_metas(meta, vendas))
```
