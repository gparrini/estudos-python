```python
from selenium import webdriver

navegador = webdriver.Chrome()
```

```python
navegador.get("file:///C:/Users/joaol/Downloads/AulaIFrame/pagina.html")
navegador.maximize_window()
```

```python
navegador.find_element("id", "campo_busca_principal").send_keys("Python")
```

```python
# entrar no iframe
elemento_iframe = navegador.find_element("id", "newsletter_iframe")

navegador.switch_to.frame(elemento_iframe)
navegador.find_element("id", "nome").send_keys("Lira")
```

```python
# voltar para a página principal
navegador.switch_to.default_content()
navegador.find_element("id", "campo_busca_principal").send_keys("Power BI")
```

```python
# identificar existência de iframes

lista_iframes = navegador.find_elements("tag name", "iframe")
print(len(lista_iframes))
```
