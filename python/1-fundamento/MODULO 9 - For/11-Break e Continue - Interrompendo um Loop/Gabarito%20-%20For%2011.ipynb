# Formas de interromper um for

### 2 Opções:

- break -> interrompe e finaliza o for
- continue -> interrompe e vai para o próximo item do for

```python
vendas = [100, 150, 1500, 2000, 120]
```

- Caso 1: Se todas as vendas forem acima da meta, a loja ganha bônus

```python
meta = 110

for venda in vendas:
    if venda < meta:
        print('A loja não ganha bônus')
        break
    print(venda)

    
```

- Caso 2: Exiba quem bateu a meta

```python
vendedores = ['João', 'Julia', 'Ana', 'José', 'Maria']
meta = 130

for venda in vendas:
    if venda < meta:
        continue
    print(venda)
```

```python
'''
DIFERENÇA ENTRE BREAK E CONTINUE NO FOR:

1. BREAK (Parada total):
   - O que faz: Encerra o loop 'for' imediatamente.
   - Analogia: Você está procurando suas chaves na gaveta. Achou? 
     Você para de procurar e fecha a gaveta (break). Não faz sentido 
     continuar olhando as outras coisas.
   - Uso comum: Quando você só precisa achar UMA coisa para tomar uma 
     decisão (ex: verificar se alguém foi reprovado, se a loja não 
     bateu a meta).

2. CONTINUE (Pula a vez):
   - O que faz: Ignora o código que está abaixo dele e volta para o 
     início do 'for' para pegar o próximo item da lista.
   - Analogia: Você está separando maçãs boas das podres. Pegou uma podre? 
     Você a descarta (continue) e pega a próxima maçã. Você não joga a 
     cesta inteira fora.
   - Uso comum: Quando você quer filtrar ou ignorar itens indesejados 
     dentro de uma lista e processar apenas os itens válidos.
'''
```
